```python
df1=pd.DataFrame({
    'id':[1,2,3,4],
    'name':['Swap','Shru','Raj','Esh']
})
df2=pd.DataFrame({
    'id':[1,2,3,4],
    'name':['Deb','Pri','Rum','Haim']
})
inner_result=pd.merge(df1,df2,how='inner',on='id')
inner_result
```

```python
df3=pd.DataFrame({
    'id':[1,2,3,4],
    'name':['Swap','Shru','Raj','Esh']
})
df4=pd.DataFrame({
    'id':[5,6,7,8],
    'name':['Deb','Pri','Rum','Haim']
})
inner_result=pd.merge(df3,df4,how='outer',on='id')
inner_result
```

```python
left_result=pd.merge(df3,df4,how='left',on='id')
left_result
```

```python
right_result=pd.merge(df3,df4,how='right',on='id')
right_result
```

```python
df5=pd.DataFrame({
    'id':[1,2,3,4],
    'name':['Swap','Shru','Raj','Esh']
})
df6=pd.DataFrame({
    'id':[5,6,7,8],
    'name':['Deb','Pri','Rum','Haim']
})
result=pd.concat([df5,df6],axis=0)
result
```
